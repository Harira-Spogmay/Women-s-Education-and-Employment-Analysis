<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 1 &nbsp;|&nbsp; Creating the Applicant Record</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">Every loan application stores the same details, so we build an <code>Applicant</code> class. Each applicant is an object created from it.</p>
</div>

```python
class Applicant:
    """Stores all information about one loan application."""

    def __init__(self, app_id, name, income, loan_amount, credit_history, debts, stable_job):
        self.app_id = app_id                  # unique ID, e.g. 101
        self.name = name                      # applicant name
        self.income = income                  # monthly income
        self.loan_amount = loan_amount        # loan requested
        self.credit_history = credit_history  # True = good, False = bad
        self.debts = debts                    # number of existing loans
        self.stable_job = stable_job          # True or False
        self.score = 0                        # calculated in Step 2
        self.decision = "Pending"             # Approved / Review / Rejected

    def show(self):
        print(f"ID: {self.app_id} | Name: {self.name} | Income: {self.income} | "
              f"Loan: {self.loan_amount} | Score: {self.score} | Decision: {self.decision}")


# Test the class
a1 = Applicant(101, "Ali", 80000, 300000, True, 1, True)
a1.show()
```

<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 2 &nbsp;|&nbsp; Loan Scoring and Decision</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">Each applicant gets a score. The score decides if the loan is Approved, Review, or Rejected.</p>
</div>

```python
def calculate_score(applicant):
    score = 0

    if applicant.credit_history:
        score += 40
    if applicant.income >= 50000:
        score += 20
    if applicant.loan_amount <= applicant.income * 5:
        score += 20
    if applicant.stable_job:
        score += 10

    score -= applicant.debts * 10
    return max(score, 0)


def predict_decision(score):
    if score >= 60:
        return "Approved"
    elif score >= 40:
        return "Review"
    else:
        return "Rejected"


def evaluate(applicant):
    applicant.score = calculate_score(applicant)
    applicant.decision = predict_decision(applicant.score)


# Test with 3 applicants
applicants = [
    Applicant(101, "Ali",   80000, 300000, True,  1, True),
    Applicant(102, "Sara",  30000, 400000, False, 3, True),
    Applicant(103, "Usman", 45000, 300000, True,  1, True),
]

for person in applicants:
    evaluate(person)
    person.show()
```

<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 3 &nbsp;|&nbsp; Application Queue (FIFO)</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">Applications are processed in the order they arrive, just like a customer line at a bank. Adding and removing both take O(1) time.</p>
</div>

```python
from collections import deque

# Create an empty queue
application_queue = deque()

# Enqueue: new applications join at the back of the line
application_queue.append(Applicant(201, "Hina",  90000, 400000, True,  0, True))
application_queue.append(Applicant(202, "Bilal", 35000, 500000, False, 2, False))
application_queue.append(Applicant(203, "Zara",  40000, 300000, True,  1, True))

print("Applications waiting in queue:", len(application_queue))
print()
print("Processing in arrival order...")

# Dequeue: the first application in line is served first
while application_queue:
    person = application_queue.popleft()
    evaluate(person)
    person.show()

print()
print("Applications left in queue:", len(application_queue))
```

<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 4 &nbsp;|&nbsp; Applicant Search (Hash Table)</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">Applicants are stored in a hash table (dictionary) with the applicant ID as the key. Searching by ID takes O(1) time, however many applicants there are.</p>
</div>

```python
# Create an empty hash table (dictionary)
applicant_table = {}

# Insert: the applicant ID is the key, the Applicant object is the value
def add_applicant(applicant):
    evaluate(applicant)
    applicant_table[applicant.app_id] = applicant

# Search: find an applicant directly by ID
def search_applicant(app_id):
    if app_id in applicant_table:
        print("Applicant found:")
        applicant_table[app_id].show()
    else:
        print(f"No applicant found with ID {app_id}")


# Add 5 applicants
add_applicant(Applicant(301, "Ayesha", 95000, 450000, True,  0, True))
add_applicant(Applicant(302, "Hamza",  38000, 600000, False, 2, True))
add_applicant(Applicant(303, "Noor",   52000, 200000, True,  1, True))
add_applicant(Applicant(304, "Faisal", 28000, 350000, False, 3, False))
add_applicant(Applicant(305, "Maryam", 60000, 250000, True,  0, True))

print("Total applicants stored:", len(applicant_table))
print()

# Search tests
search_applicant(303)
print()
search_applicant(999)
```

<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 5 &nbsp;|&nbsp; Priority Processing (Heap)</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">A priority queue serves the most important application first, not the earliest one. Here, the applicant with the highest score is processed first. Insert and remove take O(log n) time.</p>
</div>

```python
import heapq

# Create an empty priority queue (heap)
priority_queue = []

# Push the 5 applicants from Step 4 into the heap
# Python's heap serves the SMALLEST value first,
# so we store the score as a negative number to get the HIGHEST score first.
# 'order' is a tie-breaker: equal scores are served in arrival order.
for order, person in enumerate(applicant_table.values()):
    heapq.heappush(priority_queue, (-person.score, order, person))

print("Applicants in priority queue:", len(priority_queue))
print()
print("Processing by priority (highest score first)...")

# Pop: always removes the applicant with the highest score
rank = 1
while priority_queue:
    neg_score, order, person = heapq.heappop(priority_queue)
    print(f"Rank {rank}: ", end="")
    person.show()
    rank += 1
```

<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 6 &nbsp;|&nbsp; Applicant Ranking (Merge Sort)</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">Merge Sort splits the list into halves, sorts each half, and merges them back in order. It ranks applicants by score in O(n log n) time.</p>
</div>

```python
def merge(left, right):
    """Merge two sorted lists into one sorted list (highest score first)."""
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i].score >= right[j].score:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result


def merge_sort(applicants_list):
    """Sort applicants by score, highest first. Time complexity: O(n log n)."""
    if len(applicants_list) <= 1:
        return applicants_list

    mid = len(applicants_list) // 2
    left = merge_sort(applicants_list[:mid])
    right = merge_sort(applicants_list[mid:])
    return merge(left, right)


# Rank the 5 applicants stored in the hash table (Step 4)
ranked_list = merge_sort(list(applicant_table.values()))

print("Applicant Ranking (highest score first)")
print("-" * 40)
for position, person in enumerate(ranked_list, start=1):
    print(f"#{position}  {person.name:<8} Score: {person.score:<3} {person.decision}")
```

<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 7 &nbsp;|&nbsp; Fast Lookup (Binary Search)</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">Binary Search finds an applicant in a sorted list by repeatedly checking the middle and discarding half of the remaining data. It runs in O(log n) time.</p>
</div>

```python
def binary_search(sorted_list, target_id):
    """Find an applicant by ID in a list sorted by ID. Time complexity: O(log n)."""
    low = 0
    high = len(sorted_list) - 1
    steps = 0

    while low <= high:
        steps += 1
        mid = (low + high) // 2
        mid_id = sorted_list[mid].app_id

        if mid_id == target_id:
            return sorted_list[mid], steps
        elif mid_id < target_id:
            low = mid + 1      # target is in the right half
        else:
            high = mid - 1     # target is in the left half

    return None, steps


# Binary Search needs sorted data: sort the applicants by ID (smallest first)
sorted_by_id = sorted(applicant_table.values(), key=lambda a: a.app_id)

print("Applicant IDs in sorted order:", [a.app_id for a in sorted_by_id])
print()

# Search tests
for target in [303, 304, 999]:
    result, steps = binary_search(sorted_by_id, target)
    if result:
        print(f"Searching ID {target}: found in {steps} step(s)")
        result.show()
    else:
        print(f"Searching ID {target}: not found after {steps} step(s)")
    print()
```

<div style="background: linear-gradient(90deg, #0D47A1, #2196F3); color: white; padding: 18px 25px; border-radius: 10px; font-family: Arial, sans-serif;">
  <h2 style="margin: 0;">Step 8 &nbsp;|&nbsp; Loan Management System (Final Menu)</h2>
  <p style="margin: 6px 0 0 0; color: #E3F2FD;">All components are combined into one interactive system: add applicants, search by ID, view rankings, and process applications by priority.</p>
</div>

```python
import heapq

def get_number(message):
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Please enter a valid number.")

def get_yes_no(message):
    while True:
        answer = input(message + " (y/n): ").strip().lower()
        if answer in ("y", "n"):
            return answer == "y"
        print("Please type y or n.")


def menu_add():
    app_id = get_number("Applicant ID: ")
    if app_id in applicant_table:
        print("This ID already exists. Please use a different ID.")
        return
    name = input("Name: ").strip()
    income = get_number("Monthly income: ")
    loan = get_number("Loan amount requested: ")
    credit = get_yes_no("Good credit history?")
    debts = get_number("Number of existing loans: ")
    job = get_yes_no("Stable job?")

    person = Applicant(app_id, name, income, loan, credit, debts, job)
    add_applicant(person)
    print("\nApplicant added and evaluated:")
    person.show()


def menu_search_hash():
    app_id = get_number("Enter applicant ID to search: ")
    search_applicant(app_id)


def menu_search_binary():
    app_id = get_number("Enter applicant ID to search: ")
    sorted_by_id = sorted(applicant_table.values(), key=lambda a: a.app_id)
    result, steps = binary_search(sorted_by_id, app_id)
    if result:
        print(f"Found in {steps} step(s):")
        result.show()
    else:
        print(f"No applicant found with ID {app_id} (checked in {steps} step(s))")


def menu_ranking():
    ranked = merge_sort(list(applicant_table.values()))
    print("\nApplicant Ranking (highest score first)")
    print("-" * 45)
    for position, person in enumerate(ranked, start=1):
        print(f"#{position}  {person.name:<10} Score: {person.score:<3} {person.decision}")


def menu_priority():
    heap = []
    for order, person in enumerate(applicant_table.values()):
        heapq.heappush(heap, (-person.score, order, person))
    print("\nProcessing by priority (highest score first)")
    print("-" * 45)
    rank = 1
    while heap:
        _, _, person = heapq.heappop(heap)
        print(f"Rank {rank}: ", end="")
        person.show()
        rank += 1


def main_menu():
    while True:
        print("\n" + "=" * 45)
        print("      BANK LOAN MANAGEMENT SYSTEM")
        print("=" * 45)
        print("1. Add new applicant")
        print("2. Search applicant by ID (Hash Table)")
        print("3. Search applicant by ID (Binary Search)")
        print("4. Show applicant ranking (Merge Sort)")
        print("5. Process by priority (Heap)")
        print("6. Exit")
        print("-" * 45)

        choice = input("Choose an option (1-6): ").strip()

        if choice == "1":
            menu_add()
        elif choice == "2":
            menu_search_hash()
        elif choice == "3":
            menu_search_binary()
        elif choice == "4":
            menu_ranking()
        elif choice == "5":
            menu_priority()
        elif choice == "6":
            print("\nThank you for using the Bank Loan Management System.")
            break
        else:
            print("Invalid choice. Please enter a number from 1 to 6.")


main_menu()
```
