# Women’s Education and Employment Analysis

## Introduction

This project analyzes the relationship between women's education and employment.
The purpose of this project is to explore whether education level is associated
with employment status and average wages.

The project uses Python, Pandas, and Matplotlib to organize, clean, analyze,
and visualize the data. The results are presented through charts and written
interpretations to communicate important data-driven insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Education_Level": [
        "Primary", "Primary", "Secondary", "Secondary",
        "Bachelor", "Bachelor", "Master", "Master"
    ],
    "Employment_Status": [
        "Employed", "Unemployed", "Employed", "Unemployed",
        "Employed", "Employed", "Employed", "Employed"
    ],
    "Employment_Rate": [
        35, 40, 50, 55, 70, 75, 85, 90
    ],
    "Average_Wage": [
        250, 280, 350, 370, 500, 550, 700, 750
    ]
}

df = pd.DataFrame(data)

df
```

### **Analyze the Data* :



First we will Find average employment rate by education level *

```python
education_analysis = df.groupby("Education_Level")["Employment_Rate"].mean()

education_analysis
```

Find average wage by education level:

```python
wage_analysis = df.groupby("Education_Level")["Average_Wage"].mean()

wage_analysis
```

Count employment status:

```python
employment_status = df["Employment_Status"].value_counts()

employment_status
```

### **Create Your First Visualization**

Employment Rate by Education Level

```python
education_analysis.plot(kind="bar")

plt.title("Average Employment Rate by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Average Employment Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The chart compares the average employment rate across different education levels.
The results show differences in employment rates between education groups.
This visualization helps us understand the pattern between education level
and employment.

# **Create Your Second Visualization**

Create Your Second Visualization

```python
wage_analysis.plot(kind="bar")

plt.title("Average Wage by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Average Wage")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The chart shows the average wage for each education level.
The visualization allows us to compare wage differences across education groups
and observe whether higher education levels are associated with higher average wages
in this dataset.

### **Create Your Third Visualization**

Employment Status Distribution

```python
employment_status.plot(kind="pie", autopct="%1.1f%%")

plt.title("Employment Status Distribution")
plt.ylabel("")
plt.show()
```

### Interpretation

The pie chart shows the distribution of employed and unemployed individuals
in the dataset. It provides a simple overview of the employment status of
the women included in the dataset.
